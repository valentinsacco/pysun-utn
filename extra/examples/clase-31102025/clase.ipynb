```python
tabla['Temp de celda (°C)'] = tabla['Temperatura (°C)'] + 0.031 * tabla['Irradiancia (W/m²)']
```

```python
tabla
```

```python
N = 12 # Cantidad de Paneles
Gstd = 1000 # W/m2
Ppico = 240 # W
kp = -0.0044 # 1/ºC
Tr = 25
eta = 0.95
```

```python
tabla['Pot (kW)'] = N * tabla['Irradiancia (W/m²)'] / Gstd * Ppico * (1 + kp * (tabla['Temp de celda (°C)'] - Tr)) * eta * 1e-3
```

```python
tabla
```

```python
enero = tabla.loc['2019-01', ['Temp de celda (°C)', 'Pot (kW)']]
```
